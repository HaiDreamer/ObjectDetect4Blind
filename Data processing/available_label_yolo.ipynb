```python
# 1) Path
DATASET_DIR = "/content/drive/MyDrive/Available_label"
PROJECT = "/content/drive/MyDrive/Available_label"
RUN_NAME = "predict_selected_labels"
```

```python
# 2) CHECK WHETHER THE DATASET CONTAINS IMAGES
import os, re

assert os.path.isdir(DATASET_DIR), f"Directory not found: {DATASET_DIR}"

img_re = re.compile(r"\.(jpg|jpeg|png|bmp|tif|tiff|webp|HEIC)$", re.I)
n_imgs = 0

for root, _, files in os.walk(DATASET_DIR):
    for f in files:
        if img_re.search(f):
            n_imgs += 1

print("Number of images found:", n_imgs)

assert n_imgs > 0, "DATASET_DIR contains no images => YOLO will not generate any output."

```

```python
# 3) Create folder PROJECT
from pathlib import Path
Path(PROJECT).mkdir(parents=True, exist_ok=True)
```

```python
# 4) RUN PREDICTION — consume the generator when using stream=True
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

wanted = ["person", "bicycle", "car", "motorcycle", "bus", "truck", "traffic light"]
name_to_id = {n: i for i, n in model.names.items()}
wanted_ids = [name_to_id[n] for n in wanted if n in name_to_id]

gen = model.predict(
    source=DATASET_DIR,
    classes=wanted_ids,
    conf=0.1,
    imgsz=1280,
    device=0,
    save=True,
    save_txt=True,
    project=PROJECT,
    name=RUN_NAME,
    exist_ok=True,
    stream=True,
    verbose=True
)
```

```python
# 5) ITERATE OVER THE GENERATOR TO FORCE YOLO TO RUN & CREATE OUTPUT DIRECTORIES
last_res = None
for r in gen:
    last_res = r
    pass
```
