```python

import os
import yaml

DATASET_PATH = '/content/drive/MyDrive/GroupProject_Seg'  #path dataset


print(f"Dataset path: {DATASET_PATH}")
```

```python

# 1) folder want to check (can check train or test folder)
labels_train_dir = Path("/content/drive/MyDrive/GroupProject_Seg/labels/val")
print("Check folder:", labels_train_dir)
print("Exist? ", labels_train_dir.exists())

# 2) FILE TXT
detect_files = []   # only bbox (5 columns)
seg_files = []      # segmentation (>=7 columnt, (n-5) even)
mixed_files = []    # 1 file has both 5 column and seg
empty_files = []    # emty file
broken_files = []   # undefined column

for txt in labels_train_dir.rglob("*.txt"):
    with open(txt, "r") as f:
        lines = [l.strip() for l in f.readlines() if l.strip()]

    if len(lines) == 0:
        empty_files.append(str(txt))
        continue

    has_5 = False
    has_seg = False
    bad_line = False

    for line in lines:
        parts = line.split()
        n = len(parts)

        if n == 5:
            has_5 = True           #  bbox form
        elif n > 5 and (n - 5) % 2 == 0:
            has_seg = True         # seg form (polygon)
        else:
            bad_line = True        # strange column
            break

    if bad_line:
        broken_files.append(str(txt))
    elif has_5 and not has_seg:
        detect_files.append(str(txt))
    elif has_seg and not has_5:
        seg_files.append(str(txt))
    elif has_5 and has_seg:
        mixed_files.append(str(txt))

print("\n FILE DETECTION (only bbox – 5 cột)")
print("Total:", len(detect_files))
for p in detect_files[:50]:
    print(p)

print("\n FILE SEGMENTATION (only polygon)")
print("Total:", len(seg_files))
for p in seg_files[:50]:
    print(p)

print("\n  FILE MIXED (both bbox and seg) ")
print("Total:", len(mixed_files))
for p in mixed_files[:50]:
    print(p)

print("\n FILE EMPTY LABEL ")
print("Total:", len(empty_files))
for p in empty_files[:50]:
    print(p)

print("\n===== FILE WRONG FORMAT  (undefined) =====")
print("Total:", len(broken_files))
for p in broken_files[:50]:
    print(p)
```
